# Regulatory Data Analysis with Python and pandas

This notebook uses Python and pandas to explore and analyse the regulatory dataset created for my SQL portfolio project.

The analysis includes data inspection, filtering, sorting, grouping, aggregation and date-based analysis.


## 1. Load the dataset

```python
import pandas as pd
df = pd.read_csv("/content/regulatory_documents.csv")
df.head()
```

```python
df.columns
df.dtypes
df["publication_date"] = pd.to_datetime(df["publication_date"])
df["effective_date"] = pd.to_datetime(df["effective_date"])
df.dtypes
```

## 2. Inspect the data

```python
df.columns
df.dtypes
```

```python
df["publication_date"] = pd.to_datetime(df["publication_date"])
df["effective_date"] = pd.to_datetime(df["effective_date"])

df.dtypes
```

## 3. Filter and sort regulatory documents

```python
filtered_df = df[df["publication_date"] < "2025-01-01"]
print(filtered_df)
```

```python
print(df[df["publication_date"]>="2025-01-01"])
```

```python
recent_df = df[df["publication_date"] >= "2025-01-01"]
print(recent_df[["publisher_name", "title", "publication_date"]])
```

## 4. Regulatory document analysis by publisher, jurisdiction and status

```python
temp_1 = (df.groupby("publisher_name")["document_id"].count())
print(temp_1.sort_values(ascending=False))
```

```python
jurisdiction_counts = df.groupby("jurisdiction_name")["document_id"].count()
print(jurisdiction_counts.sort_values(ascending=False))
```

```python
status_counts = df.groupby("status")["document_id"].count()
print(status_counts)
```
